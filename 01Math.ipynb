```python
import math

valorFlotante = float(input("numero con decimal: "))

print(f"El siguiente entero hacia abajo de {valorFlotante} es: {math.floor(valorFlotante)}")

print(f"El siguiente entero hacia arriba de {valorFlotante} es: {math.ceil(valorFlotante)}")

print(f"La parte entera truncada de {valorFlotante} es {math.trunc(valorFlotante)}")
```

```python
numero = int(input("Numero ENTERO: "))

print(f"La raiz cuadrada de {numero} es igual a {math.sqrt(numero)}")

print(f"El factorial de {numero} es igual a {math.factorial(numero)}")
```

```python
potencia = int(input("Dame un numero entero: "))

print(f"El resultado al elevar {numero} a la {potencia}a potencia es igual a {math.pow(numero, potencia)}")

print(f"El valor de Pi es {math.pi}")
```
