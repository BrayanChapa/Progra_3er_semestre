```python
from collections import namedtuple

Punto = namedtuple("Punto", "x y")
punto = Punto(10, 20)
```

```python
print(f"Ubicación del punto sobre el eje X = {punto.x}")
print(f"Ubicación del punto sobre el eje Y = {punto.y}")
```

```python
Persona = namedtuple("Persona", ("nombres", "primer_apellido", "segundo_apellido"),
         defaults=["N/A"])
mi_persona = Persona("Juan", "Dominguez")

print(f"El nombre de la persona es: {mi_persona.nombres}")
print(f"El primer apellido de la persona es: {mi_persona.primer_apellido}")
print(f"El segundo apellido de la persona es: {mi_persona.segundo_apellido}")
```

```python
tupla_ordinaria = (50, 45)
mi_punto = Punto._make(tupla_ordinaria)

print(mi_punto)
```
