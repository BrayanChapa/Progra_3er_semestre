```python
lista_numeros = [1,2,3,4,5,6]
lista_letras = ("a","e","i","o","u", "z", "x", "y")
print(lista_numeros)
print(lista_letras)

resultado_combinado = zip(lista_numeros, lista_letras)
print(f"\nEl tipo del objeto combinado resultante es {type(resultado_combinado)}")

resultado_en_lista = list(resultado_combinado)
print(f"El resultado combinado es: {resultado_en_lista}")
```

```python
lista_numeros = [1,2,3,4,5,6]
def al_doble(valor1):
    return valor1 + valor1

numeros_al_doble = map(al_doble, lista_numeros)


resultado_en_lista = list(numeros_al_doble)
print(f"El doble de cada elemento de la lista original : {resultado_en_lista}")
```
