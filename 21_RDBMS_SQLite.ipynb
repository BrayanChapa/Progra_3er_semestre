```python
import sys
import sqlite3
from sqlite3 import Error

try:
  with sqlite3.connect("PrimerIntentXDemo.db") as conn:
    mi_cursor = conn.cursor()
    mi_cursor.execute("SELECT * FROM proyecto ORDER BY nombre")
    registros = mi_cursor.fetchall()

    if registros:
      print("Claves\tNombre")
      print("*" * 30)
      for clave, nombre in registros:
        print(f"{clave}\t{nombre}")
    else:
      print("No se encontraron registros")
except Error as e:
  print(e)
except Exception:
  print(f"Se produjo el siguiente error: {sys.exc_info()[0]}")
finally:
  conn.close
```

```python
import sys
import sqlite3
from sqlite3 import Error

valor_clave = int(input("Ingrese la clave: "))

try:
  with sqlite3.connect("PrimerIntentXDemo.db") as conn:
    mi_cursor = conn.cursor()
    valores = {"clave":valor_clave}
    mi_cursor.execute(f"SELECT * FROM proyecto WHERE clave = :clave", valores)
    registro = mi_cursor.fetchall()

    if registro:
      for clave, nombre in registro:
        print(f"Clave: {clave}\tNombre: {nombre}")
    else:
      print("No se encontraron registros")

except Error as e:
  print(e)
except Exception:
  print(f"Se produjo el siguiente error: {sys.exc_info()[0]}")
finally:
  conn.close
```

```python
import sys
import sqlite3

try:
  with sqlite3.connect("ConFechas_timestamp.db") as conexion:
    mi_cursor = conexion.cursor()
    intruccion = "CREATE TABLE amigo (clave INTEGER PRIMARY KEY, nombre TEXT NOT NULL, fecha_registro TIMESTAMP);"
    mi_cursor.execute(intruccion)
    print("Tabla creada")
except sqlite3.Error as e:
  print(e)
except:
  print(f"Se produjo el siguiente error: {sys.exc_info()[0]}")
finally:
  if(conexion):
    conexion.close()
    print("Se cerro la conexion")
```

```python
import sys
import sqlite3
import datetime
from sqlite3 import Error

try:
  with sqlite3.connect("ConFechas_timestamp.db") as conn:
    mi_cursor = conn.cursor()
    fecha_actual = datetime.datetime.combine(datetime.date.today(), datetime.datetime.min.time())
    valores = {"clave":30, "nombre":"Crescencio", "fecha_registro":fecha_actual}
    mi_cursor.execute("INSERT INTO amigo VALUES(:clave, :nombre, :fecha_registro)", valores)
    print("Registro insertado exitosamente")
except Error as e:
  print(e)
except:
  print(f"Se produjo el siguiente error: {sys.exc_info()[0]}")
finally:
  if (conn):
    conn.close
```

```python
import sqlite3
from sqlite3 import Error
import sys

try:
  with sqlite3.connect("ConFechas_timestamp.db", detect_type = sqlite3.PARSE_DECLTYPES | sqlite3.PARSE_COLNAMES) as conn:
    mi_cursor = conn.cursor()
    mi_cursor.execute("SELECT clave, nombre, fecha_registro FROM amigo")
    registros = mi_cursor.fetchall()

    if registros:
      for clave, nombre, fecha_registro in registros:
        print(f"Clave: {clave}\tNombre: {nombre}\tFecha: {fecha_registro}")
    else:
      print("No se encontraron registros")

except Error as e:
  print(e)
except:
  print(f"Se produjo el siguiente error: {sys.exc_info()[0]}")
finally:
  conn.close
```
