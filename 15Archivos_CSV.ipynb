```python
import csv
import sys
datos_a_leer = dict()

try:
  with open("autos.csv", "r", encoding="latin1", newline= "") as archivo:
    lector = csv.reader(archivo)
    next(lector)

    for clave, marca, año in lector:
      datos_a_leer[int(clave)] = (marca, int(año))
except FileNotFoundError:
  print("El archivo no se encontro")
except Exception:
  Exception = sys.exc_info()
  print("Ocurrio un problema de tipo",Exception)
else:
  print(datos_a_leer)
```

```python
import sys
datos_a_leer = dict()

try:
  with open("autos.csv", "r", encoding="latin1", newline= "") as archivo:
    lector = csv.reader(archivo)
    next(lector)

    for clave, marca, año in lector:
      datos_a_leer[int(clave)] = (marca, int(año))
except FileNotFoundError:
  print("El archivo no se encontro")
except csv.Error as fallo_csv:
  print("Fallo al leer el archivo CSV", fallo_csv)
except Exception:
  Exception = sys.exc_info()
  print("Ocurrio un problema de tipo",Exception)
else:
  print(datos_a_leer)
```

```python
Entradas = [
    ["correo", "nombre", "telefono"],
    ["@gmail.com", "Juan", "67"],
    ["maria@gmail.com","maria","6767"],
    ["diana@gmail.com","diana","6767667"]
]

for e in Entradas:
  print(f"{e[1]}|{e[2]}|{e[3]}")
```

```python
print(">> Contenido de la lista.\n")
print(Entradas)

if os.path.exists("agenda.csv"):
  if (os.path.exists("agenda.bak")):
    os.remove("agenda.bak")
    os.rename("agenda.csv", "agenda.bak")

with open("agenda.csv", "w+") as f:
  for e in Entradas:
    f.write(f"{e[1]},{e[2]},{e[3]}")

print("\n>> Contenido del archivo.\n")
with open("agenda.csv") as f:
  print(f.read())
```

```python
Datos = []

with open("agenda.csv","r") as f:
  for linea in f:
    lista = linea.split(",")
    lista[2] = lista[2].replace("\n","")
    Datos.append(lista)

print(Datos)
```
