```python
conjunto_uno = {"Hugo", "Paco", "Luis"}
conjunto_dos= {"Britni", "Llojan", "Luis"}
print("CONJUNTOS")
print(conjunto_uno)
print(conjunto_dos)
print()

print("UNIÓN")
print(conjunto_uno | conjunto_dos)
print()

print("INTERSECCIÓN")
print(conjunto_uno & conjunto_dos)
print()

print("DIFERENCIA")
print(conjunto_uno - conjunto_dos)
print()

print("DIFERENCIA SIMËTRICA")
print(conjunto_uno ^ conjunto_dos)
```

```python
conjunto_tres = {"Paco", "Luis"}

print(conjunto_tres <= conjunto_uno)
```
